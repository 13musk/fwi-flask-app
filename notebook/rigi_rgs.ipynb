```python
import pandas as pd
import numpy as np
import matplotlib as plt
import seaborn as sns
%matplotlib inline 
```

```python
dataset=pd.read_csv('Algerian_forest_fires_dataset_UPDATE.csv',header=1)
```

```python
dataset.head()
```

```python
dataset.info()
```

### Data Cleaning

```python
##missing values
dataset.isnull().any(axis=1)
```

```python
dataset[dataset.isnull().any(axis=1)]
```

 the dataset is converted into two sets based on Region from 122th index,we can make a new column based on the region

 1 :"Bejaia region Dataset"

 2: "sidi -Bel Abbes Region Dataset"

 Add new column with region

```python
dataset.loc[:122,'Region']=0
dataset.loc[122:,'Region']=1
```

```python
dataset.info()
```

```python
df=dataset
```

```python
df[['Region']]=df[['Region']].astype('int')
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
## Remove the null values
df=df.dropna().reset_index(drop=True)
df.head()
```

```python
df.isnull().sum()
```

```python
df.iloc[[122]]
```

```python
#remove 122 row
df=df.drop(122).reset_index(drop=True)
```

```python
df.iloc[[122]]
```

```python
df.columns
```

```python
#fix spaces in coloumn name
df.columns=df.columns.str.strip()
df.columns
```

```python
df.info()
```

## Change the required column as interger data type

```python
df.columns
```

```python
df[['month','day','year','Temperature','RH','Ws']]=df[['month','day','year','Temperature','RH','Ws']].astype(int)
```

```python
df.info()
```

## changing the other columns into flote data type


```python
objects=[features for features in df.columns if df[features].dtypes=='str']
```

```python
for i in objects:
    if i!='Classes':
        df[i]=df[i].astype(float)
```

```python
df.info()
```

```python
df.describe
```

```python
df.head()
```

```python
##lets save the clean dataset
df.to_csv("lgerian_forest_fires_clean_dataset.csv",index='False')
```
